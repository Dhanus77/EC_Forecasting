```python
df_eda = df.copy()

print(df_eda.shape)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(15, 5))

plt.plot(df_eda["date"], df_eda["Appliances"])

plt.title("Appliance Energy Consumption Over Time")
plt.xlabel("Date")
plt.ylabel("Energy Consumption (Wh)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df_eda["hour"] = df_eda["date"].dt.hour

hourly_consumption = df_eda.groupby("hour")["Appliances"].mean()

print(hourly_consumption)
```

```python
plt.figure(figsize=(10, 5))

plt.plot(hourly_consumption.index, hourly_consumption.values, marker="o")

plt.title("Average Appliance Energy Consumption by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Average Consumption (Wh)")
plt.xticks(range(24))
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
df_eda["day_of_week"] = df_eda["date"].dt.day_name()

day_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

daily_consumption = (
    df_eda.groupby("day_of_week")["Appliances"]
    .mean()
    .reindex(day_order)
)

print(daily_consumption)
```

```python
plt.figure(figsize=(10, 5))

plt.bar(daily_consumption.index, daily_consumption.values)

plt.title("Average Appliance Energy Consumption by Day")
plt.xlabel("Day of Week")
plt.ylabel("Average Consumption (Wh)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
weather_columns = [
    "Appliances",
    "T_out",
    "RH_out",
    "Windspeed",
    "Visibility",
    "Tdewpoint",
    "Press_mm_hg"
]

weather_correlation = df_eda[weather_columns].corr()["Appliances"].sort_values(
    ascending=False
)

print(weather_correlation)
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    df_eda["T_out"],
    df_eda["Appliances"],
    alpha=0.3
)

plt.title("Outdoor Temperature vs Appliance Energy Consumption")
plt.xlabel("Outdoor Temperature (°C)")
plt.ylabel("Appliance Energy Consumption (Wh)")
plt.tight_layout()
plt.show()
```

```python
df_eda["is_weekend"] = df_eda["date"].dt.dayofweek >= 5

weekend_consumption = (
    df_eda.groupby("is_weekend")["Appliances"]
    .mean()
)

print(weekend_consumption)
```

```python
plt.figure(figsize=(7, 5))

plt.bar(
    ["Weekday", "Weekend"],
    weekend_consumption.values
)

plt.title("Average Appliance Energy Consumption: Weekday vs Weekend")
plt.ylabel("Average Consumption (Wh)")
plt.tight_layout()
plt.show()
```
